```python
# import libraries
import pandas as pd
from sklearn.linear_model import LogisticRegression
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score
```

```python
#load dataset
iris_data = pd.read_csv('iris.csv')
iris_data.head()
```

```python
#split dataset into features and labels
x = iris_data.drop(columns=['variety'])
y = iris_data['variety']
```

```python
x.head()
```

```python
# split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#standardize the features
scaler = StandardScaler()
x_train_scaled  = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# create a model and train without the training set
model_unscaled = LogisticRegression(max_iter=200)
model_unscaled.fit(x.values,y)
```

```python
# make prediciton with the trained model
predictions = model_unscaled.predict([[4.6, 3.5, 1.5, 0.2]])
print(predictions)
print("Predicted class:", predictions[0])
```

```python
#create a model and train the model with scaled features
model = LogisticRegression(max_iter=200)
model.fit(x_train_scaled, y_train)
```

```python
# make prediciton with the trained model using scaled features
predictions = model.predict([[4.6, 3.5, 1.5, 0.2]])
print(predictions)
print("Predicted class:", predictions[0])
```

```python
#evaluate the model
y_pred = model.predict(x_test_scaled)
accuracy = accuracy_score(y_test, y_pred)
print(f"Model accuracy: {accuracy * 100:.2f}%")
```

```python
# create sample data for predictions
new_data = np.array([
          [5.1,3.5,1.4,0.2],
          [6.3,2.9,5.6,1.8],
          [4.9,3.0,1.4,0.2]])
```

```python
predictions = model.predict(new_data)
print(predictions)
```
