```python
# import libraries
import pandas as pd
from sklearn.linear_model import LogisticRegression
```

```python

# load a dataset
iris_data = pd.read_csv('iris.csv')
#iris_data = pd.read_csv('https://gist.githubusercontent.com/netj/8836201/raw/6f9306ad21398ea43cba4f7d537619d0e07d5ae3/iris.csv')
# show the columns in the dataset
print(f"columns in dataset are: {iris_data.columns}")
# check dataset with head method to show first five rows
iris_data.head()
```

```python
#split dataset into features and labels
x = iris_data.drop(columns=['variety'])
y= iris_data['variety']

# show the features for training the model
print(f"features for training the model are: {x.columns}")
# show the labels for training the model
print(f"labels for training the model are: {y.name}")
```

```python
#show the actual labels
print(f"actual labels are: {y.unique()}")
#show the first five rows of features
print("first five rows of features:")
x.head()
```

```python
# create a ml  regression model
model = LogisticRegression()
```

```python
#train the model
model.fit(x.values, y)
```

```python
#predict the labels using the trained model
predictions = model.predict([[4.6, 3.5, 1.5, 0.2]])
print(f"predicted label for input [4.6, 3.5, 1.5, 0.2] is: {predictions[0]}")

print(predictions)
```
